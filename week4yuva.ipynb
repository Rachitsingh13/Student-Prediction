```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate,
    GridSearchCV
)

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    PolynomialFeatures
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression

from sklearn.svm import SVC

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

from sklearn.inspection import permutation_importance
import joblib
FILE_PATH = "C:/Users/HP/Downloads/collegePlace.csv"

df = pd.read_csv(FILE_PATH)


print("=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
print(df.head())
print("\nColumn names:")

for column in df.columns:
    print(column)

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

print(df.dtypes)

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()

print(
    "Shape after removing duplicates:",
    df.shape
)

print("\n" + "=" * 70)
print("TARGET DISTRIBUTION")
print("=" * 70)

print(
    df["PlacedOrNot"].value_counts()
)


print("\nTarget percentages:")

print(
    df["PlacedOrNot"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="PlacedOrNot"
)

plt.title(
    "Placement Status Distribution"
)

plt.xlabel(
    "Placed or Not Placed"
)

plt.ylabel(
    "Number of Students"
)

plt.xticks(
    [0, 1],
    ["Not Placed", "Placed"]
)

plt.show()
X = df.drop(
    columns=["PlacedOrNot"]
)
y = df["PlacedOrNot"]
print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)
# ============================================================
# 10. DEFINE FEATURE TYPES
# ============================================================
# Numerical columns

numeric_features = [
    "Age",
    "Internships",
    "CGPA",
    "Hostel",
    "HistoryOfBacklogs"
]


# Categorical columns

categorical_features = [
    "Gender",
    "Stream"
]


print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 11. CHECK WHETHER ALL REQUIRED COLUMNS EXIST
# ============================================================

required_columns = (
    numeric_features
    + categorical_features
    + ["PlacedOrNot"]
)


missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]


if len(missing_columns) > 0:

    raise ValueError(
        "Missing columns: "
        + str(missing_columns)
    )


print("\nAll required columns are present.")


# ============================================================
# 12. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print("\n" + "=" * 70)
print("TRAIN TEST SPLIT")
print("=" * 70)

print(
    "Training samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)


# ============================================================
# 13. PREPROCESSING
# ============================================================

# Numerical preprocessing:
#
# Missing values
#       ↓
# Median
#       ↓
# StandardScaler

numeric_pipeline = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),

    (
        "scaler",
        StandardScaler()
    )
])


# Categorical preprocessing:
#
# Missing values
#       ↓
# Most frequent
#       ↓
# One Hot Encoding

categorical_pipeline = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),

    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])


preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),

        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)


# ============================================================
# 14. STRATIFIED 5-FOLD CROSS VALIDATION
# ============================================================

cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=42
)


# ============================================================
# 15. BASELINE MODELS
# ============================================================

baseline_models = {

    "Logistic Regression":

        Pipeline([

            (
                "preprocessor",
                preprocessor
            ),

            (
                "classifier",
                LogisticRegression(
                    max_iter=2000,
                    random_state=42
                )
            )
        ]),


    "SVM":

        Pipeline([

            (
                "preprocessor",
                preprocessor
            ),

            (
                "classifier",
                SVC(
                    kernel="rbf",
                    probability=True,
                    random_state=42
                )
            )
        ]),


    "Random Forest":

        Pipeline([

            (
                "preprocessor",
                preprocessor
            ),

            (
                "classifier",
                RandomForestClassifier(
                    n_estimators=100,
                    random_state=42
                )
            )
        ])
}


# ============================================================
# 16. BASELINE MODEL EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("BASELINE MODEL RESULTS")
print("=" * 70)


baseline_results = []


for name, model in baseline_models.items():

    scores = cross_validate(

        model,

        X_train,

        y_train,

        cv=cv,

        scoring=[
            "accuracy",
            "precision",
            "recall",
            "f1",
            "roc_auc"
        ],

        n_jobs=-1
    )


    baseline_results.append({

        "Model": name,

        "Accuracy":
        scores["test_accuracy"].mean(),

        "Precision":
        scores["test_precision"].mean(),

        "Recall":
        scores["test_recall"].mean(),

        "F1":
        scores["test_f1"].mean(),

        "ROC-AUC":
        scores["test_roc_auc"].mean()
    })


baseline_results_df = pd.DataFrame(
    baseline_results
)


print(
    baseline_results_df.round(4)
)


# ============================================================
# 17. FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 70)
print("FEATURE ENGINEERING")
print("=" * 70)


# We create interaction features between numerical
# variables.
#
# Examples:
#
# Age × CGPA
# Age × Internships
# CGPA × Internships
# CGPA × Hostel
# etc.


numeric_interaction_pipeline = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),

    (
        "interaction",
        PolynomialFeatures(
            degree=2,

            interaction_only=True,

            include_bias=False
        )
    ),

    (
        "scaler",
        StandardScaler()
    )
])


categorical_engineered_pipeline = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),

    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])


engineered_preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",
            numeric_interaction_pipeline,
            numeric_features
        ),

        (
            "categorical",
            categorical_engineered_pipeline,
            categorical_features
        )
    ]
)


# ============================================================
# 18. LOGISTIC REGRESSION WITH FEATURE ENGINEERING
# ============================================================

lr_engineered = Pipeline([

    (
        "preprocessor",
        engineered_preprocessor
    ),

    (
        "classifier",
        LogisticRegression(
            max_iter=3000,
            random_state=42
        )
    )
])


# ============================================================
# 19. EVALUATE FEATURE ENGINEERED LOGISTIC REGRESSION
# ============================================================

scores = cross_validate(

    lr_engineered,

    X_train,

    y_train,

    cv=cv,

    scoring=[
        "accuracy",
        "precision",
        "recall",
        "f1",
        "roc_auc"
    ],

    n_jobs=-1
)


feature_engineering_result = {

    "Model":
    "Logistic Regression + Interactions",

    "Accuracy":
    scores["test_accuracy"].mean(),

    "Precision":
    scores["test_precision"].mean(),

    "Recall":
    scores["test_recall"].mean(),

    "F1":
    scores["test_f1"].mean(),

    "ROC-AUC":
    scores["test_roc_auc"].mean()
}


print("\nFeature engineered model:")

print(
    pd.DataFrame(
        [feature_engineering_result]
    ).round(4)
)


# ============================================================
# 20. RANDOM FOREST HYPERPARAMETER TUNING
# ============================================================

print("\n" + "=" * 70)
print("RANDOM FOREST HYPERPARAMETER TUNING")
print("=" * 70)


rf_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "classifier",
        RandomForestClassifier(
            random_state=42
        )
    )
])


rf_parameters = {

    "classifier__n_estimators":
    [100, 200, 300],

    "classifier__max_depth":
    [None, 5, 10, 15],

    "classifier__min_samples_split":
    [2, 5, 10],

    "classifier__min_samples_leaf":
    [1, 2, 4],

    "classifier__max_features":
    ["sqrt", "log2"]
}


rf_grid = GridSearchCV(

    estimator=rf_pipeline,

    param_grid=rf_parameters,

    scoring="f1",

    cv=cv,

    n_jobs=-1,

    verbose=1,

    return_train_score=True
)


rf_grid.fit(
    X_train,
    y_train
)


print("\nBest Random Forest parameters:")

print(
    rf_grid.best_params_
)


print("\nBest Random Forest CV F1:")

print(
    round(
        rf_grid.best_score_,
        4
    )
)


# ============================================================
# 21. SVM HYPERPARAMETER TUNING
# ============================================================

print("\n" + "=" * 70)
print("SVM HYPERPARAMETER TUNING")
print("=" * 70)


svm_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "classifier",
        SVC(
            probability=True,
            random_state=42
        )
    )
])


svm_parameters = {

    "classifier__C":
    [0.1, 1, 10, 100],

    "classifier__gamma":
    ["scale", "auto", 0.01, 0.1, 1],

    "classifier__kernel":
    ["rbf"]
}


svm_grid = GridSearchCV(

    estimator=svm_pipeline,

    param_grid=svm_parameters,

    scoring="f1",

    cv=cv,

    n_jobs=-1,

    verbose=1
)


svm_grid.fit(
    X_train,
    y_train
)


print("\nBest SVM parameters:")

print(
    svm_grid.best_params_
)


print("\nBest SVM CV F1:")

print(
    round(
        svm_grid.best_score_,
        4
    )
)


# ============================================================
# 22. LOGISTIC REGRESSION HYPERPARAMETER TUNING
# ============================================================

print("\n" + "=" * 70)
print("LOGISTIC REGRESSION HYPERPARAMETER TUNING")
print("=" * 70)


lr_pipeline = Pipeline([

    (
        "preprocessor",
        engineered_preprocessor
    ),

    (
        "classifier",
        LogisticRegression(
            max_iter=3000,
            random_state=42
        )
    )
])


lr_parameters = {

    "classifier__C":
    [0.01, 0.1, 1, 10, 100],

    "classifier__solver":
    ["liblinear", "lbfgs"],

    "classifier__class_weight":
    [None, "balanced"]
}


lr_grid = GridSearchCV(

    estimator=lr_pipeline,

    param_grid=lr_parameters,

    scoring="f1",

    cv=cv,

    n_jobs=-1,

    verbose=1
)


lr_grid.fit(
    X_train,
    y_train
)


print("\nBest Logistic Regression parameters:")

print(
    lr_grid.best_params_
)


print("\nBest Logistic Regression CV F1:")

print(
    round(
        lr_grid.best_score_,
        4
    )
)


# ============================================================
# 23. GRADIENT BOOSTING
# ============================================================

print("\n" + "=" * 70)
print("GRADIENT BOOSTING HYPERPARAMETER TUNING")
print("=" * 70)


gb_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "classifier",
        GradientBoostingClassifier(
            random_state=42
        )
    )
])


gb_parameters = {

    "classifier__n_estimators":
    [50, 100, 200],

    "classifier__learning_rate":
    [0.01, 0.05, 0.1],

    "classifier__max_depth":
    [2, 3, 5],

    "classifier__min_samples_split":
    [2, 5, 10]
}


gb_grid = GridSearchCV(

    estimator=gb_pipeline,

    param_grid=gb_parameters,

    scoring="f1",

    cv=cv,

    n_jobs=-1,

    verbose=1
)


gb_grid.fit(
    X_train,
    y_train
)


print("\nBest Gradient Boosting parameters:")

print(
    gb_grid.best_params_
)


print("\nBest Gradient Boosting CV F1:")

print(
    round(
        gb_grid.best_score_,
        4
    )
)


# ============================================================
# 24. CREATE FINAL MODEL DICTIONARY
# ============================================================

tuned_models = {

    "Tuned Logistic Regression":
    lr_grid.best_estimator_,

    "Tuned SVM":
    svm_grid.best_estimator_,

    "Tuned Random Forest":
    rf_grid.best_estimator_,

    "Tuned Gradient Boosting":
    gb_grid.best_estimator_
}


# ============================================================
# 25. FINAL TEST SET EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)


final_results = []


for name, model in tuned_models.items():

    # Train

    model.fit(
        X_train,
        y_train
    )


    # Predict

    y_pred = model.predict(
        X_test
    )


    # Probability

    if hasattr(
        model,
        "predict_proba"
    ):

        y_prob = (
            model
            .predict_proba(X_test)
            [:, 1]
        )

    else:

        y_prob = (
            model
            .decision_function(X_test)
        )


    # Metrics

    accuracy = accuracy_score(
        y_test,
        y_pred
    )


    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )


    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )


    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )


    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )


    final_results.append({

        "Model": name,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1": f1,

        "ROC-AUC": roc_auc
    })


final_results_df = pd.DataFrame(
    final_results
)


final_results_df = (
    final_results_df
    .sort_values(
        by="F1",
        ascending=False
    )
    .reset_index(drop=True)
)


print(
    final_results_df.round(4)
)


# ============================================================
# 26. SELECT BEST MODEL
# ============================================================

best_model_name = (
    final_results_df
    .iloc[0]["Model"]
)


best_model = tuned_models[
    best_model_name
]


print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print(
    "Best model:",
    best_model_name
)


# ============================================================
# 27. FINAL MODEL TRAINING
# ============================================================

best_model.fit(
    X_train,
    y_train
)


# ============================================================
# 28. FINAL PREDICTIONS
# ============================================================

y_pred = best_model.predict(
    X_test
)


if hasattr(
    best_model,
    "predict_proba"
):

    y_prob = (
        best_model
        .predict_proba(X_test)
        [:, 1]
    )

else:

    y_prob = (
        best_model
        .decision_function(X_test)
    )


# ============================================================
# 29. FINAL CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL CLASSIFICATION REPORT")
print("=" * 70)


print(
    classification_report(

        y_test,

        y_pred,

        target_names=[
            "Not Placed",
            "Placed"
        ],

        zero_division=0
    )
)


# ============================================================
# 30. FINAL METRICS
# ============================================================

final_accuracy = accuracy_score(
    y_test,
    y_pred
)


final_precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)


final_recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)


final_f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)


final_auc = roc_auc_score(
    y_test,
    y_prob
)


print("\nFinal Accuracy :", round(final_accuracy, 4))

print("Final Precision:", round(final_precision, 4))

print("Final Recall   :", round(final_recall, 4))

print("Final F1 Score :", round(final_f1, 4))

print("Final ROC-AUC  :", round(final_auc, 4))


# ============================================================
# 31. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)


print("\n" + "=" * 70)
print("CONFUSION MATRIX")
print("=" * 70)

print(cm)


plt.figure(
    figsize=(6, 5)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=[
        "Not Placed",
        "Placed"
    ],

    yticklabels=[
        "Not Placed",
        "Placed"
    ]
)


plt.title(
    "Confusion Matrix"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.tight_layout()

plt.show()


# ============================================================
# 32. TRUE/FALSE POSITIVE/NEGATIVE
# ============================================================

tn, fp, fn, tp = cm.ravel()


print("\nTrue Negative :", tn)

print("False Positive:", fp)

print("False Negative:", fn)

print("True Positive :", tp)


# ============================================================
# 33. FALSE POSITIVE ANALYSIS
# ============================================================

false_positive_mask = (

    (y_test == 0)

    &

    (y_pred == 1)
)


false_positives = X_test[
    false_positive_mask
].copy()


false_positives["Actual"] = 0

false_positives["Predicted"] = 1


print("\n" + "=" * 70)
print("FALSE POSITIVE STUDENTS")
print("=" * 70)


print(
    false_positives
)


# ============================================================
# 34. FALSE NEGATIVE ANALYSIS
# ============================================================

false_negative_mask = (

    (y_test == 1)

    &

    (y_pred == 0)
)


false_negatives = X_test[
    false_negative_mask
].copy()


false_negatives["Actual"] = 1

false_negatives["Predicted"] = 0


print("\n" + "=" * 70)
print("FALSE NEGATIVE STUDENTS")
print("=" * 70)


print(
    false_negatives
)


# ============================================================
# 35. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(

    y_test,

    y_prob
)


plt.figure(
    figsize=(7, 6)
)


plt.plot(

    fpr,

    tpr,

    label=f"AUC = {final_auc:.3f}"
)


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"
)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    f"ROC Curve - {best_model_name}"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 36. PERMUTATION FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE")
print("=" * 70)


importance = permutation_importance(

    best_model,

    X_test,

    y_test,

    scoring="f1",

    n_repeats=10,

    random_state=42,

    n_jobs=-1
)


importance_df = pd.DataFrame({

    "Feature":
    X_test.columns,

    "Importance":
    importance.importances_mean

})


importance_df = (
    importance_df
    .sort_values(
        by="Importance",
        ascending=False
    )
)


print(
    importance_df
)


# ============================================================
# 37. TOP FEATURES VISUALIZATION
# ============================================================

top_features = (
    importance_df
    .head(10)
)


plt.figure(
    figsize=(9, 6)
)


sns.barplot(

    data=top_features,

    x="Importance",

    y="Feature"
)


plt.title(
    "Top 10 Important Features"
)

plt.tight_layout()

plt.show()


# ============================================================
# 38. BASELINE RANDOM FOREST VS FINAL MODEL
# ============================================================

baseline_rf = (
    baseline_results_df[
        baseline_results_df["Model"]
        == "Random Forest"
    ]
    .iloc[0]
)


final_best = final_results_df.iloc[0]


comparison_df = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Precision",

        "Recall",

        "F1",

        "ROC-AUC"
    ],


    "Baseline Random Forest": [

        baseline_rf["Accuracy"],

        baseline_rf["Precision"],

        baseline_rf["Recall"],

        baseline_rf["F1"],

        baseline_rf["ROC-AUC"]
    ],


    "Final Tuned Model": [

        final_best["Accuracy"],

        final_best["Precision"],

        final_best["Recall"],

        final_best["F1"],

        final_best["ROC-AUC"]
    ]
})


comparison_df["Improvement"] = (

    comparison_df["Final Tuned Model"]

    -

    comparison_df["Baseline Random Forest"]
)


print("\n" + "=" * 70)
print("BASELINE VS FINAL MODEL")
print("=" * 70)


print(
    comparison_df.round(4)
)


# ============================================================
# 39. MODEL PERFORMANCE GRAPH
# ============================================================

plot_data = final_results_df.set_index(
    "Model"
)[
    [
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "ROC-AUC"
    ]
]


plot_data.plot(

    kind="bar",

    figsize=(12, 6)
)


plt.title(
    "Final Model Performance Comparison"
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Model"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()


# ============================================================
# 40. SAVE RESULTS
# ============================================================

baseline_results_df.to_csv(
    "baseline_results.csv",
    index=False
)


final_results_df.to_csv(
    "final_model_results.csv",
    index=False
)


importance_df.to_csv(
    "feature_importance.csv",
    index=False
)


comparison_df.to_csv(
    "baseline_vs_final.csv",
    index=False
)


false_positives.to_csv(
    "false_positives.csv",
    index=False
)


false_negatives.to_csv(
    "false_negatives.csv",
    index=False
)


# ============================================================
# 41. SAVE BEST MODEL
# ============================================================

joblib.dump(

    best_model,

    "best_student_placement_model.pkl"
)


print("\nBest model saved successfully.")


# ============================================================
# 42. PREDICTION FUNCTION FOR NEW STUDENT
# ============================================================

def predict_student(
    age,
    gender,
    stream,
    internships,
    cgpa,
    hostel,
    history_of_backlogs
):

    new_student = pd.DataFrame({

        "Age": [age],

        "Gender": [gender],

        "Stream": [stream],

        "Internships": [internships],

        "CGPA": [cgpa],

        "Hostel": [hostel],

        "HistoryOfBacklogs": [
            history_of_backlogs
        ]
    })


    prediction = best_model.predict(
        new_student
    )[0]


    probability = (
        best_model
        .predict_proba(new_student)[0][1]
    )


    if prediction == 1:

        result = "Placed"

    else:

        result = "Not Placed"


    print("\n" + "=" * 50)

    print("STUDENT PREDICTION")

    print("=" * 50)

    print(
        "Prediction:",
        result
    )

    print(
        f"Placement Probability: "
        f"{probability * 100:.2f}%"
    )


    return result, probability


# ============================================================
# 43. EXAMPLE PREDICTION
# ============================================================

# Example student:
#
# Age = 22
# Gender = Male
# Stream = Computer Science
# Internships = 2
# CGPA = 8
# Hostel = 1
# HistoryOfBacklogs = 0


# Uncomment to test:

# predict_student(
#     age=22,
#     gender="Male",
#     stream="Computer Science",
#     internships=2,
#     cgpa=8,
#     hostel=1,
#     history_of_backlogs=0
# )


# ============================================================
# 44. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("WEEK 4 PROJECT COMPLETED")
print("=" * 70)

print(
    "Best Model:",
    best_model_name
)

print(
    "Accuracy:",
    round(final_accuracy, 4)
)

print(
    "Precision:",
    round(final_precision, 4)
)

print(
    "Recall:",
    round(final_recall, 4)
)

print(
    "F1 Score:",
    round(final_f1, 4)
)

print(
    "ROC-AUC:",
    round(final_auc, 4)
)

print("\nGenerated files:")

print("1. baseline_results.csv")
print("2. final_model_results.csv")
print("3. feature_importance.csv")
print("4. baseline_vs_final.csv")
print("5. false_positives.csv")
print("6. false_negatives.csv")
print("7. best_student_placement_model.pkl")

print("\nAll experiments completed successfully.")
```
